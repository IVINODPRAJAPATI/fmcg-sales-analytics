```python
import pandas as pd
df=pd.read_csv('FMCG_2022_2024.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Negative stock_available:",(df['stock_available']<0).sum())
print("Negative delivery_qty:",(df['delivered_qty']<0).sum())
```

```python
print("Negative units_sold:", (df['units_sold'] < 0).sum())
```

```python
df = df[(df['stock_available'] >= 0) & (df['delivered_qty'] >= 0) & (df['units_sold'] >= 0)]
print(df.shape)
```

```python
df.duplicated().sum()
```

```python
df['date']=pd.to_datetime(df['date'])
```

```python
df['month']=df['date'].dt.month
```

```python
df['quarter']=df['date'].dt.quarter
```

```python
df['day_of_week']=df['date'].dt.dayofweek
```
